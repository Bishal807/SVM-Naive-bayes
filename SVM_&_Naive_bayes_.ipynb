### Theoretical Questions and Answers

1. **What is a Support Vector Machine (SVM)?**

   * SVM is a supervised learning algorithm used for classification and regression tasks. It finds the hyperplane that best separates the data into classes with the maximum margin.

2. **Difference between Hard Margin and Soft Margin SVM?**

   * Hard Margin SVM assumes perfectly separable data without misclassification.
   * Soft Margin SVM allows some misclassifications to handle non-linearly separable data by introducing a penalty.

3. **Mathematical intuition behind SVM?**

   * SVM aims to maximize the margin between support vectors and the decision boundary, solving a quadratic optimization problem subject to constraints.

4. **Role of Lagrange Multipliers in SVM?**

   * Lagrange Multipliers help solve the constrained optimization problem by converting it into a dual problem that is easier to solve.

5. **What are Support Vectors in SVM?**

   * These are data points that lie closest to the decision boundary and influence its position and orientation.

6. **What is a Support Vector Classifier (SVC)?**

   * It's the SVM model used specifically for classification tasks.

7. **What is a Support Vector Regressor (SVR)?**

   * SVR adapts the principles of SVM for regression, using a margin of tolerance (ε) within which no penalty is given.

8. **Kernel Trick in SVM?**

   * It allows SVM to operate in a high-dimensional space without explicitly computing the coordinates, by using kernel functions.

9. **Compare Linear, Polynomial, and RBF Kernels.**

   * Linear: Best for linearly separable data.
   * Polynomial: Captures interactions up to a specified degree.
   * RBF: Handles non-linear relationships using Gaussian functions.

10. **Effect of the C parameter in SVM?**

* C controls the trade-off between achieving a low training error and a low testing error. A large C means fewer misclassifications.

11. **Role of the Gamma parameter in RBF Kernel SVM?**

* Gamma defines how far the influence of a single training example reaches; low values mean far, high values mean close.

12. **What is the Naïve Bayes classifier, and why is it called "Naïve"?**

* It is a probabilistic classifier based on Bayes' theorem assuming feature independence (the naïve assumption).

13. **What is Bayes' Theorem?**

* $P(A|B) = \frac{P(B|A)P(A)}{P(B)}$ - calculates the probability of hypothesis A given data B.

14. **Differences between Gaussian, Multinomial, and Bernoulli Naïve Bayes:**

* Gaussian: For continuous data.
* Multinomial: For count data (e.g., word counts).
* Bernoulli: For binary features.

15. **When to use Gaussian Naïve Bayes?**

* When the features are continuous and assumed to be normally distributed.

16. **Key assumptions made by Naïve Bayes?**

* All features are conditionally independent given the class label.

17. **Advantages and disadvantages of Naïve Bayes?**

* Advantages: Fast, simple, works well with text.
* Disadvantages: Assumes feature independence, which is rarely true in real data.

18. **Why is Naïve Bayes good for text classification?**

* Text data often fits the assumptions (independent word features), and the model is efficient for large datasets.

19. **Compare SVM and Naïve Bayes for classification tasks.**

* SVM: Better with high-dimensional, non-linear data.
* Naïve Bayes: Simpler, faster, works well for text classification.

20. **How does Laplace Smoothing help in Naïve Bayes?**

* It prevents zero probabilities by adding a small constant to frequency counts.

```python
# Q27: Train SVM with different C values and compare decision boundaries
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_classification
from sklearn.svm import SVC
from mlxtend.plotting import plot_decision_regions

X, y = make_classification(n_features=2, n_redundant=0, n_informative=2, n_clusters_per_class=1, n_samples=100, random_state=42)
fig, axs = plt.subplots(1, 3, figsize=(15, 4))

for i, C in enumerate([0.1, 1, 10]):
    clf = SVC(kernel='poly', degree=3, C=C)
    clf.fit(X, y)
    plt.sca(axs[i])
    plot_decision_regions(X, y, clf=clf, legend=2)
    plt.title(f'C = {C}')
plt.tight_layout()
plt.show()
```

```python
#40
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
import numpy as np

def custom_kernel(X, Y):
    return (X @ Y.T) ** 2 + 1  # Example: inhomogeneous polynomial kernel

X, y = make_classification(n_samples=100, n_features=2, n_classes=2,
                           n_informative=2, n_redundant=0, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

clf = SVC(kernel=custom_kernel)
clf.fit(X_train, y_train)
print("Custom kernel SVM accuracy:", accuracy_score(y_test, clf.predict(X_test)))
```
